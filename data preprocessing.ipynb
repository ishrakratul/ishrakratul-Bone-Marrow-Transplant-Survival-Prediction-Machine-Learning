```python
import pandas as pd
import numpy as np
```

```python
data1=pd.read_csv ("csv_result-bone-marrow-uci.csv")
data2=data1.drop('id',axis=1)
```

```python
data3=data2.replace(to_replace='?', value= None, inplace=False, limit=None, regex=False, method='pad')
```

```python
data3["recipient_body_mass"] = pd.to_numeric(data3["recipient_body_mass"])
data3["CD3_x1e8_per_kg"] = pd.to_numeric(data3["CD3_x1e8_per_kg"])
data3["CD3_to_CD34_ratio"] = pd.to_numeric(data3["CD3_to_CD34_ratio"])
```

```python
data3["recipient_body_mass"].fillna(int(data3["recipient_body_mass"].mean()), inplace=True)
data3["CD3_x1e8_per_kg"].fillna(int(data3["CD3_x1e8_per_kg"].mean()), inplace=True)
data3["CD3_to_CD34_ratio"].fillna(int(data3["CD3_to_CD34_ratio"].mean()), inplace=True)
```

```python
from feature_engine.imputation import CategoricalImputer
imputer = CategoricalImputer(imputation_method='frequent', variables=None, 
                             return_object=False, ignore_format=False)
imputer.fit(data3)
data4= imputer.transform(data3)
```

```python
data=pd.get_dummies(data4, prefix=None, prefix_sep='_', 
               dummy_na=False, columns=None, sparse=False, 
               drop_first=True)
```

```python
import pandas_profiling
from pandas_profiling import ProfileReport
ProfileReport(data, title="Pandas Profiling Report", explorative=True)
```
