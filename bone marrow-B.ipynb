```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn import metrics
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression

from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier

from sklearn.ensemble import GradientBoostingClassifier
from sklearn.ensemble import AdaBoostClassifier
from xgboost import XGBClassifier

from sklearn.metrics import confusion_matrix

from sklearn.preprocessing import StandardScaler
```

```python
x_train2=pd.read_csv ("x_train_all.csv")
x_train1=x_train2.drop('sl',axis=1)

y_train2=pd.read_csv ("y_train_all.csv")
y_train=y_train2.drop('sl',axis=1)

x_test2=pd.read_csv ("x_test_all.csv")
x_test1=x_test2.drop('sl',axis=1)

y_test2=pd.read_csv ("y_test_all.csv")
y_test=y_test2.drop('sl',axis=1)

```

```python
scale = StandardScaler()
x_train=scale.fit_transform(x_train1)

x_test=scale.fit_transform(x_test1)

```

```python
#x_train.shape

#y_train.shape

#x_test.shape

#y_test.shape
```

```python
model=DecisionTreeClassifier(criterion='gini', max_depth=7, min_samples_leaf=9, min_samples_split=5, splitter='random')

model.fit(x_train,y_train)

y_pred=model.predict(x_test)

met = confusion_matrix(y_test, y_pred)

print(met)

tp=met[0][0]
fp=met[0][1]
fn=met[1][0]
tn=met[1][1]
pr=tp/(tp+fp)
rec=tp/(tp+fn)
f1=2*pr*rec/(pr+rec)
acc=(tp+tn)/(tp+fp+tn+fn)
print("DecisionTree accuracy:",acc)
print("DecisionTree precision:",pr)
print("DecisionTree f1 score:",f1)
print("DecisionTree recall:",rec)

```

```python
model1=RandomForestClassifier(bootstrap= True, min_samples_leaf= 2, min_samples_split= 20, n_estimators= 50)

model1.fit(x_train,y_train)

y1_pred = model1.predict(x_test)

met = confusion_matrix(y_test, y1_pred)

print(met)

tp=met[0][0]
fp=met[0][1]
fn=met[1][0]
tn=met[1][1]
pr=tp/(tp+fp)
rec=tp/(tp+fn)
f1=2*pr*rec/(pr+rec)
acc=(tp+tn)/(tp+fp+tn+fn)
print("RFC accuracy:",acc)
print("RFC precision:",pr)
print("RFC f1 score:",f1)
print("RFC recall:",rec)
```

```python
model2=LogisticRegression(C= 784.7599703514607, max_iter= 10, penalty='l1', solver= 'liblinear')

model2.fit(x_train,y_train)

y2_pred = model2.predict(x_test)

met = confusion_matrix(y_test, y2_pred)
print(met)

tp=met[0][0]
fp=met[0][1]
fn=met[1][0]
tn=met[1][1]
pr=tp/(tp+fp)
rec=tp/(tp+fn)
f1=2*pr*rec/(pr+rec)
acc=(tp+tn)/(tp+fp+tn+fn)
print("LR accuracy:",acc)
print("LR precision:",pr)
print("LR f1 score:",f1)
print("LR recall:",rec)
```

```python
model3=KNeighborsClassifier(leaf_size= 2, n_neighbors= 10, p= 1, weights= 'distance')

model3.fit(x_train,y_train)

y3_pred = model3.predict(x_test)

met = confusion_matrix(y_test, y3_pred)
print(met)

tp=met[0][0]
fp=met[0][1]
fn=met[1][0]
tn=met[1][1]
pr=tp/(tp+fp)
rec=tp/(tp+fn)
f1=2*pr*rec/(pr+rec)
acc=(tp+tn)/(tp+fp+tn+fn)
print("KNN accuracy:",acc)
print("KNN precision:",pr)
print("KNN f1 score:",f1)
print("KNN recall:",rec)
```

```python
model4=GradientBoostingClassifier(learning_rate= 0.001, max_depth= 1, n_estimators= 100)
model4.fit(x_train,y_train)

y4_pred = model4.predict(x_test)

met = confusion_matrix(y_test, y4_pred)
print(met)

tp=met[0][0]
fp=met[0][1]
fn=met[1][0]
tn=met[1][1]
pr=tp/(tp+fp)
rec=tp/(tp+fn)
f1=2*pr*rec/(pr+rec)
acc=(tp+tn)/(tp+fp+tn+fn)
print("Gradient boosting classifier accuracy:",acc)
print("Gradient boosting classifierC precision:",pr)
print("Gradient boosting classifier f1 score:",f1)
print("Gradient boosting classifier recall:",rec)
```

```python
model5=AdaBoostClassifier(learning_rate=0.05, n_estimators= 50)
model5.fit(x_train,y_train)

y5_pred = model5.predict(x_test)

met = confusion_matrix(y_test, y5_pred)
print(met)

tp=met[0][0]
fp=met[0][1]
fn=met[1][0]
tn=met[1][1]
pr=tp/(tp+fp)
rec=tp/(tp+fn)
f1=2*pr*rec/(pr+rec)
acc=(tp+tn)/(tp+fp+tn+fn)
print("AdaBoost accuracy:",acc)
print("AdaBoost precision:",pr)
print("AdaBoost f1 score:",f1)
print("AdaBoost recall:",rec)
```

```python
model6=XGBClassifier(colsample_bytree= 0.5,gamma= 0, learning_rate= 0.1, max_depth= 1, min_child_weight= 0, n_estimators= 100)
model6.fit(x_train,y_train)

y6_pred = model6.predict(x_test)

met = confusion_matrix(y_test, y6_pred)
print(met)

tp=met[0][0]
fp=met[0][1]
fn=met[1][0]
tn=met[1][1]
pr=tp/(tp+fp)
rec=tp/(tp+fn)
f1=2*pr*rec/(pr+rec)
acc=(tp+tn)/(tp+fp+tn+fn)
print("XGB accuracy:",acc)
print("XGB precision:",pr)
print("XGB f1 score:",f1)
print("XGB recall:",rec)
```
