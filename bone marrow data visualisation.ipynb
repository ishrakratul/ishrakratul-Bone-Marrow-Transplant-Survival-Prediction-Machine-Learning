```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn
```

```python
data1=pd.read_csv ("manipulated data.csv")
data=data1.drop('sl',axis=1)
```

```python
data.describe()
```

```python
data.hist(layout=(8,8), sharex=False,sharey=True,figsize=(17, 10))
#plt.tight_layout(pad=1)
#plt.subplot_tool()
plt.subplots_adjust(left=0.1,bottom=0.1, right=0.9, top=0.9, wspace=0.2, hspace=0.4)
#plt.savefig("file.tiff",dpi=120,bbox_inches='tight')
plt.show()
```

```python
data.plot(kind='density',subplots=True,layout=(8,8), sharex=False,figsize=(20, 10))
plt.subplots_adjust(left=0.1,bottom=0.1, right=0.9, top=0.9, wspace=0.3, hspace=0.3)
plt.show()
```

```python
data.plot(kind='box',subplots=True,layout=(8,8), sharex=False,figsize=(20, 10))
plt.subplots_adjust(left=0.1,bottom=0.1, right=0.9, top=0.9, wspace=0.3, hspace=0.3)
plt.show()
```

```python


data_pairplot=data.dropna()
seaborn.pairplot(data_pairplot,height=1.5)
plt.show()

```

```python
plt.figure(figsize=(14,14))
csfont = {'fontname':'Times New Roman'}
seaborn.set(font_scale=2,font='Times New Roman')
res=seaborn.heatmap(data.corr(),annot=True,cmap="magma",linewidths=3,linecolor='white',square='False',
                  cbar_kws={"shrink": 0.816},vmax=0.5, annot_kws={"fontsize":16,"fontweight":'bold',**csfont})
#res.set_xticklabels(res.get_xmajorticklabels(), fontsize = 14)
res.set_xticklabels(res.get_xmajorticklabels(), fontsize = 20, rotation=90,fontweight='bold',**csfont)
res.set_yticklabels(res.get_ymajorticklabels(), fontsize = 20, rotation=0,fontweight='bold',**csfont)

#plt.savefig("heatmap_11 features data.tiff",dpi=150,bbox_inches='tight')
plt.show()
```
