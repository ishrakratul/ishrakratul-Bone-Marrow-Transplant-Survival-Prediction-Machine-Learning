```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_digits
from sklearn.feature_selection import SelectKBest 
from sklearn.feature_selection import chi2
```

```python
data1=pd.read_csv ("manipulated data.csv")
data=data1.drop('sl',axis=1)
```

```python
data.shape
```

```python
x=data.drop('survival_status',axis=1)
y=data['survival_status']
```

```python
bestfeatures=SelectKBest(score_func=chi2,k=5)
fit=bestfeatures.fit(x,y)
dfscores=pd.DataFrame(fit.scores_)
dfcolumns=pd.DataFrame(x.columns)
featureScores=pd.concat([dfcolumns,dfscores],axis=1)
featureScores.columns=['Specs','Score']
```

```python
x_new=pd.DataFrame(bestfeatures.transform(x))
```

```python
featureScores
```

```python
x_new.describe()
```

```python
featureScores.sort_values(['Score'],ascending=False)
```

```python
featureScores.sort_values(['Score'],ascending=False).to_csv('chi squared results sorted.csv')
```
